```python
import re
import random
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
random.seed(42)


def clean_string(text):
    text = text.lower()
    # оставляем только буквы, цифры, пробелы (исправлен баг с \\s → \s)
    text = re.sub(r'[^a-z0-9\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

with open("text-autocomplete/data/tweets.txt", "r", encoding="utf-8", errors="ignore") as f:
    raw_lines = [line.strip() for line in f if line.strip()]

print(f"Сырых строк: {len(raw_lines)}")
```

```python
pd.DataFrame({"text": raw_lines}).to_csv("/home/rays/Загрузки/Projects/text-autocomplete/data/raw_dataset.csv", index=False)
```

```python
seq_len = 7

cleaned_texts = [clean_string(line) for line in raw_lines]
cleaned_texts = [t for t in cleaned_texts if len(t.split()) >= seq_len]
print(f"После очистки: {len(cleaned_texts)}")
```

```python
pd.DataFrame({"text": cleaned_texts}).to_csv("/home/rays/Загрузки/Projects/text-autocomplete/data/dataset_processed.csv", index=False)
```

```python

# --- сплит 80/10/10 ---
train_val, test_texts = train_test_split(cleaned_texts, test_size=0.1, random_state=42)
train_texts, val_texts  = train_test_split(train_val,   test_size=0.1/0.9, random_state=42)

total = len(train_texts) + len(val_texts) + len(test_texts)

print(f"Train: {len(train_texts)} ({len(train_texts)/total*100:.1f}%)")
print(f"Val:   {len(val_texts)}  ({len(val_texts)/total*100:.1f}%)")
print(f"Test:  {len(test_texts)}  ({len(test_texts)/total*100:.1f}%)")
print(f"Total: {total}")
```

```python

pd.DataFrame({"text": train_texts}).to_csv("/home/rays/Загрузки/Projects/text-autocomplete/data/train.csv", index=False)
pd.DataFrame({"text": val_texts}).to_csv("/home/rays/Загрузки/Projects/text-autocomplete/data/val.csv", index=False)
pd.DataFrame({"text": test_texts}).to_csv("/home/rays/Загрузки/Projects/text-autocomplete/data/test.csv", index=False)
```
